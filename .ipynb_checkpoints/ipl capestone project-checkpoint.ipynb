```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv('IPL 2022.csv')
df
```

```python
df.info()
```

```python
print(f"Your rows are {df.shape[0]} and Your columns are {df.shape[1]}")
```

```python
df.isnull().sum()
```

```python
# Which team won the most matches
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x = match_wins.values, palette='viridis')
plt.title("Matches Win By Each Team")
```

```python
#Toss decisions
sns.countplot(x = df['toss_decision'], palette='rainbow')
plt.title("Toss Decisions")
```

```python
# Toss winners vs Match winners
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100) / df.shape[0]
percentage.round(2)
```

```python
# How do teams wins by runs or by wickets
sns.countplot(x = df['won_by'])
plt.title("Won By")
```

```python
#Most Player of the Match
count = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x = count.values, y = count.index, palette='mako')
plt.title("Top 10 Player of the Matches")
plt.ylabel("Name of player")
plt.xlabel('Number of Player of the match')
```

```python
# 2 top scorers
high_scorer = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(2)
high_scorer
```

```python
high_scorer.plot(kind = 'barh')
```

```python
# 10 best bowling figures
df['highest_wicket'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highest_wicket'] = df['highest_wicket'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending = False).head(10)
sns.barplot(x = top_bowlers.values, y = top_bowlers.index)
```

```python
# Most matches played by Venue
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(x = venue_count.values, y = venue_count.index, palette= 'rainbow')
plt.title("Number of matches played on each venue")
plt.xlabel("Number of matches")
```

```python
# who won the highest margin by runs
df[df['won_by'] == 'Runs'].sort_values(by = 'margin', ascending = False).head(1)[['match_winner', 'margin']]
```

```python
# which player had the highest individual score
df[df['highscore'] == df['highscore'].max()][['top_scorer', 'highscore']]
```

```python
# Which figure had the best bowling figures
df[df['highest_wicket'] == df['highest_wicket'].max()][['best_bowling','best_bowling_figure']]
```
